```python
'''
melt and pivot

>> melt --> wide to long format
melt( id_vars, value_vars, var_name, value_name)
       keep  , unpivot   , new_col name, new_col val)

>> pivot --> long to wide format
pivot( index, columns, values)
      keep ,

'''

df = pd.DataFrame({
    "country" : ["USA", "india", "USA", "india"],
    "year" : [2020, 2018, 2023, 2021],
    "sales" : [100, 120, 95, 110],
    "profit" : [15, 20, 10, 17],
    })

melted_df = df.melt(
    id_vars = ["country", "year"],
    value_vars = ["sales", "profit"],
    var_name = "matrices",
    value_name = "value",
)



melted_df.pivot(
    index = ["country", "year"],
    columns = "matrices",
    values = "value"
)

    
```

```python
# DATA CONCATINATION
'''
stacking dataframe on top of each other (rows) or side by side (col) '''

df1 = pd.DataFrame({
    "id" : [1, 2, 3],
    "name" : ["akbar", "azam", "khan"]})

df2 = pd.DataFrame({
    "id" : [4, 5, 6],
    "name" : ["ali", "ahmad", "jan"]})


pd.concat([df1, df2])    # row wise   --> org labels preserved
pd.concat([df1, df2], ignore_index = True )     # labels not preserved

pd.concat([df1, df2], axis = 1)    # col wise
```
